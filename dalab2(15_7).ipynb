Task 1: Import the required Python library.

```python
import pandas as pd
```

Task 2: Load the India dataset into a DataFrame.

```python
india=pd.read_csv("hondu_india_sales.csv")
india.head()
```

Task 3: Load the Pakistan dataset into a DataFrame.

```python
pakistan=pd.read_csv("hondu_pakistan_sales.csv")
pakistan.head()
```

Task 4: Load the Bangladesh dataset into a DataFrame.

```python
bangladesh=pd.read_csv("hondu_bangladesh_sales.csv")
bangladesh.head()
```

Task 5: Add `Country_Code` to the India dataset.

```python
india["Country_Code"] = "IN"
india.head()
```

Task 6: Add `Country_Code` to the Pakistan dataset.

```python
pakistan["Country_Code"] = "PK"
pakistan.head()
```

Task 7: Add `Country_Code` to the Bangladesh dataset.

```python
bangladesh["Country_Code"] = "BD"
bangladesh.head()
```

Task 8: Rename `Customer_Rating_Out_of_5` as `Rating_Out_of_5`.

```python
india.rename(columns={"Customer_Rating_Out_of_5":"Rating_Out_of_5"}, inplace=True)
india.head()
```

Task 9: Rename `Discount_INR` as `Discount_Amount_INR`.

```python
india.rename(columns={"Discount_INR":"Discount_Amount_INR"}, inplace=True)
india.head()
```

Task 10: Create `Final_Price_INR` for the India dataset.

```python
india["Final_Price_INR"] = india["Listed_Price_INR"] - india["Discount_Amount_INR"]
india.head()
```

Task 11: Convert Pakistan listed price from PKR to INR and create `Listed_Price_INR`.

```python
pakistan["Listed_Price_INR"] = pakistan["Listed_Price_PKR"] * 0.30
pakistan.head()
```

Task 12: Convert Pakistan customer rating from out of 10 to out of 5 and create `Rating_Out_of_5`.

```python
pakistan["Rating_Out_of_5"] = pakistan["Customer_Rating_Out_of_10"] / 2
pakistan.head()
```

Task 13: Convert Pakistan percentage discount into discount amount in INR and create `Discount_Amount_INR`.

```python
pakistan["Discount_Amount_INR"] = (pakistan["Listed_Price_INR"] *pakistan["Discount_Percentage"] / 100)
pakistan.head()
```

Task 14: Create `Final_Price_INR` for the Pakistan dataset.

```python
pakistan["Final_Price_INR"] = (
    pakistan["Listed_Price_INR"]
    - pakistan["Discount_Amount_INR"])
pakistan.head()
```

Task 15: Convert Bangladesh listed price from BDT to INR and create `Listed_Price_INR`.

```python
bangladesh["Listed_Price_INR"] = bangladesh["Listed_Price_BDT"] * 0.70
bangladesh.head()
```

Task 16:Rename `Customer_Rating_Out_of_5` as `Rating_Out_of_5`.

```python
bangladesh.rename(columns={"Customer_Rating_Out_of_5":"Rating_Out_of_5"}, inplace=True)
bangladesh.head()
```

Task 17: Convert Bangladesh discount from BDT to INR and create `Discount_Amount_INR`.

```python
bangladesh["Discount_Amount_INR"] = bangladesh["Discount_BDT"] * 0.70
bangladesh.head()
```

Task 18: Create `Final_Price_INR` for the Bangladesh dataset.

```python
bangladesh["Final_Price_INR"] = bangladesh["Listed_Price_BDT"] - bangladesh["Discount_BDT"]
bangladesh.head()
```

Task 19: Keep only the required final columns in the India dataset

```python
final_columns = [
    "Sale_ID",
    "Month",
    "Country_Code",
    "Model",
    "Segment",
    "Sales_Units",
    "Listed_Price_INR",
    "Rating_Out_of_5",
    "Discount_Amount_INR",
    "Final_Price_INR"
]
india = india[final_columns]
india.head()
```

Task 20: Keep only the required final columns in the Pakistan dataset.

```python
pakistan = pakistan[final_columns]
pakistan.head()
```

Task 21: Keep only the required final columns in the Bangladesh dataset.

```python
bangladesh = bangladesh[final_columns]
bangladesh.head()
```

Task 22: Combine the three standardized national datasets into one regional dataset.

```python
regional = pd.concat(
    [india, pakistan, bangladesh],
    ignore_index=True)

regional.head()
```

Task 23: Reset the index of the final regional dataset.

```python
regional = regional.reset_index(drop=True)
regional.head()
```
